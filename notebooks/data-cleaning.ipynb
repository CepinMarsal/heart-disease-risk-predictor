# Heart Disease Risk Predictor
# Data Cleaning
## Objective
This notebook focuses on improving the dataset quality before exploratory data analysis and machine learning modeling.

The cleaning process includes:
- Converting placeholder missing values into proper NaN values.
- Correcting data types.
- Handling missing values.
- Converting the target variable into binary classification.
- Validating the cleaned dataset.
- Exporting the cleaned dataset for the next stage.

```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
```

```python
columns = [
    "age","sex","cp","trestbps","chol","fbs",
    "restecg","thalach","exang","oldpeak",
    "slope","ca","thal","target"
]

df = pd.read_csv(
    "../data/raw/processed.cleveland.data",
    names=columns
)
```

## Replace Placeholder Missing Values

```python
df.replace("?", np.nan, inplace=True)
```

# Missing Value Inspection

```python
df.isna().sum()
```

## Correct Data Types

```python
df["ca"] = pd.to_numeric(df["ca"])
df["thal"] = pd.to_numeric(df["thal"])
```

```python
df.dtypes
```

## Handle Missing Values

```python
df["ca"].fillna(df["ca"].median(), inplace=True)
df["thal"].fillna(df["thal"].median(), inplace=True)
```

## Validate Missing Values

```python
df.isna().sum()
```

## Check Duplicate Records

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

## Convert Target Variable

```python
df["target"] = df["target"].apply(lambda x: 0 if x == 0 else 1)
```

```python
df["target"].value_counts()
```

## Final Validation

```python
df.info()
```

```python
df.describe().T
```

## Export Clean Dataset

```python
df.to_csv(
    "../data/processed/heart_disease_clean.csv",
    index=False
)
```

# Cleaning Summary
The dataset has been successfully cleaned and prepared for exploratory data analysis.

Completed tasks:
- Replaced placeholder missing values (`?`) with `NaN`
- Corrected data types
- Imputed missing values using the median
- Removed duplicate records
- Converted the target variable into binary classification
- Exported the cleaned dataset for the next stage
